```python
import pandas as pd
import numpy as np
```

```python
np.random.seed(42)

data = {
    'Customer_ID': range(1, 1001),
    'Age': np.random.randint(18, 70, 1000),
    'Annual_Spend': np.random.randint(100, 5000, 1000),
    'Region': np.random.choice(
        ['North', 'South', 'East', 'West'],
        1000
    )
}

df = pd.DataFrame(data)

print("Dataset created successfully!")
print("Dataset size:", df.shape)

df.head()
```

```python
print(df.info())
```

```python
df.describe()
```

```python
bins = [0, 1000, 3000, 6000]

labels = [
    'Low Spender',
    'Medium Spender',
    'High Spender'
]

df['Spend_Segment'] = pd.cut(
    df['Annual_Spend'],
    bins=bins,
    labels=labels
)

df.head()
```

```python
print(df['Spend_Segment'].value_counts())
```

```python
print(
    df['Spend_Segment']
    .value_counts(normalize=True)
    .mul(100)
)
```

```python
print(df.head(20))
```

```python
df.to_csv(
    'segmented_customers.csv',
    index=False
)

print("Data successfully processed and exported to CSV!")
```
