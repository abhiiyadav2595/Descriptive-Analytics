```python
import pandas as pd
import numpy as np
```

```python
data = {
    "Employee_ID": [101,102,103,104,105,106,107,108,109,110],
    "Department": ["IT","HR","Finance","IT","Sales","Finance","HR","IT","Sales","Marketing"],
    "Experience_Years": [2,5,8,3,6,10,4,7,5,9],
    "Salary": [45000,52000,78000,50000,62000,90000,55000,75000,64000,85000],
    "Performance_Score": [78,85,92,80,88,95,83,90,86,94]
}
```

```python
df = pd.DataFrame(data)
```

```python
print("="*70)
print("EMPLOYEE PERFORMANCE DATASET")
print("="*70)

print(df)

print("="*70)
```

```python
print("\nDataset Shape")
print(df.shape)

print("\nColumn Names")
print(df.columns)

print("\nData Types")
print(df.dtypes)
```

```python
print("\nOverall Statistical Summary")

print(df.describe())
```

```python
print("\nMean Values")

print(df[["Experience_Years",
          "Salary",
          "Performance_Score"]].mean())
```

```python
print("\nMedian Values")

print(df[["Experience_Years",
          "Salary",
          "Performance_Score"]].median())
```

```python
print("\nMode Values")

print(df[["Experience_Years",
          "Salary",
          "Performance_Score"]].mode())
```

```python
print("\nVariance")

print(df[["Experience_Years",
          "Salary",
          "Performance_Score"]].var())
```

```python
print("\nStandard Deviation")

print(df[["Experience_Years",
          "Salary",
          "Performance_Score"]].std())
```

```python
print("\nCorrelation Matrix")

print(df[["Experience_Years",
          "Salary",
          "Performance_Score"]].corr())
```

```python
print("\nAverage Salary Department Wise")

print(df.groupby("Department")["Salary"].mean())
```

```python
print("\nAverage Performance Department Wise")

print(df.groupby("Department")["Performance_Score"].mean())
```

```python
print("\nLowest 3 Salaries")

print(df.nsmallest(3,"Salary"))
```
