```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = {
    "Month": ["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug"],
    "Website_Visitors": [1500,1800,2100,2400,2800,3200,3500,3900],
    "Marketing_Cost": [20000,23000,25000,27000,31000,34000,36000,40000],
    "Sales_Revenue": [85000,92000,105000,118000,132000,145000,160000,178000],
    "Product_Category": ["Laptop","Mobile","Laptop","Tablet","Mobile","Laptop","Tablet","Mobile"]
}

```

```python
df = pd.DataFrame(data)
```

```python
print("="*80)
print("ONLINE BUSINESS SALES DATASET")
print("="*80)

print(df)

print("="*80)
```

```python
sns.set_theme(style="darkgrid")
```

```python
plt.figure(figsize=(12,6))

sns.lineplot(
    data=df,
    x="Month",
    y="Sales_Revenue",
    marker="o",
    linewidth=3,
    color="green"
)

plt.title("Monthly Sales Revenue Trend",fontsize=16,fontweight="bold")
plt.xlabel("Month",fontsize=12)
plt.ylabel("Revenue (₹)",fontsize=12)
plt.grid(True)

for x,y in zip(df["Month"],df["Sales_Revenue"]):
    plt.text(x,y+2000,str(y),ha='center',fontsize=9)

plt.show()
```

```python
plt.figure(figsize=(12,6))

sns.scatterplot(
    data=df,
    x="Marketing_Cost",
    y="Sales_Revenue",
    hue="Product_Category",
    size="Website_Visitors",
    sizes=(120,350),
    palette="Set2"
)

plt.title("Marketing Cost vs Sales Revenue",fontsize=16,fontweight="bold")
plt.xlabel("Marketing Cost (₹)")
plt.ylabel("Sales Revenue (₹)")
plt.grid(True)

plt.show()
```

```python
category_sales = df.groupby("Product_Category",as_index=False)["Sales_Revenue"].mean()

plt.figure(figsize=(10,6))

sns.barplot(
    data=category_sales,
    x="Product_Category",
    y="Sales_Revenue",
    palette="viridis"
)

plt.title("Average Revenue by Product Category",fontsize=16,fontweight="bold")
plt.xlabel("Product Category")
plt.ylabel("Average Revenue (₹)")

for index,row in category_sales.iterrows():
    plt.text(index,row["Sales_Revenue"]+1000,
             int(row["Sales_Revenue"]),
             ha='center',
             fontsize=10)

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    df[["Website_Visitors","Marketing_Cost","Sales_Revenue"]].corr(),
    annot=True,
    cmap="YlGnBu",
    linewidths=1,
    square=True
)

plt.title("Correlation Matrix",fontsize=15,fontweight="bold")

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.histplot(
    df["Sales_Revenue"],
    bins=8,
    kde=True,
    color="orange"
)

plt.title("Sales Revenue Distribution",fontsize=15,fontweight="bold")
plt.xlabel("Revenue")
plt.ylabel("Frequency")

plt.show()
```
