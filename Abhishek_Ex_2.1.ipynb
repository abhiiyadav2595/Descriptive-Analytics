```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = {
    "Month": ["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug"],
    "Website_Visitors": [1500,1800,2100,2400,2800,3200,3500,3900],
    "Marketing_Cost": [20000,23000,25000,27000,31000,34000,36000,40000],
    "Sales_Revenue": [85000,92000,105000,118000,132000,145000,160000,178000],
    "Product_Category": ["Laptop","Mobile","Laptop","Tablet","Mobile","Laptop","Tablet","Mobile"]
}

```

```python
df = pd.DataFrame(data)
```

```python
print("="*80)
print("ONLINE BUSINESS SALES DATASET")
print("="*80)

print(df)

print("="*80)
```

```python
sns.set_theme(style="darkgrid")
```

```python
# 1. Histogram
plt.figure(figsize=(8, 5))
sns.histplot(df["Sales_Revenue"], bins=8)
plt.title("Histogram of Sales Revenue")
plt.xlabel("Sales Revenue")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.boxplot(
    data=df,
    x="Sales_Revenue",
    width=0.4
)

mean_value = df["Sales_Revenue"].mean()

plt.axvline(
    mean_value,
    linestyle="--",
    linewidth=2,
    label=f"Mean: {mean_value:.0f}"
)

plt.title("Boxplot of Sales Revenue", fontsize=15, fontweight="bold")
plt.xlabel("Sales Revenue (₹)")
plt.legend()
plt.grid(True, axis="x", alpha=0.3)

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.kdeplot(
    data=df,
    x="Sales_Revenue",
    fill=True,
    linewidth=2,
    label="Sales Revenue Density"
)

mean_value = df["Sales_Revenue"].mean()

plt.axvline(
    mean_value,
    linestyle="--",
    linewidth=2,
    label=f"Mean: {mean_value:.0f}"
)

plt.title("Density Plot of Sales Revenue", fontsize=15, fontweight="bold")
plt.xlabel("Sales Revenue (₹)")
plt.ylabel("Density")
plt.legend()
plt.grid(True, alpha=0.3)

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    df[["Website_Visitors","Marketing_Cost","Sales_Revenue"]].corr(),
    annot=True,
    cmap="YlGnBu",
    linewidths=1,
    square=True
)

plt.title("Correlation Matrix",fontsize=15,fontweight="bold")

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.histplot(
    df["Sales_Revenue"],
    bins=8,
    kde=True,
    color="orange"
)

plt.title("Sales Revenue Distribution",fontsize=15,fontweight="bold")
plt.xlabel("Revenue")
plt.ylabel("Frequency")

plt.show()
```
