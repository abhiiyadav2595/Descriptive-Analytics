```python
# ============================================================
# EXPERIMENT 3
# Binomial Probability Distribution using Coin Flip Simulation
# Author: Your Name
# ============================================================

# Step 1 : Import Libraries
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Style
sns.set_theme(style="whitegrid")

# Step 2 : Simulation Parameters

np.random.seed(42)        # Makes output reproducible

n_flips = 15              # Coin flips in one experiment
prob_head = 0.5           # Fair coin
trials = 20000            # Total experiments

# Step 3 : Simulate Coin Flips

heads = np.random.binomial(
    n=n_flips,
    p=prob_head,
    size=trials
)

# Step 4 : Experimental Statistics

experimental_mean = np.mean(heads)
experimental_variance = np.var(heads)

# Step 5 : Theoretical Statistics

theoretical_mean = n_flips * prob_head
theoretical_variance = n_flips * prob_head * (1 - prob_head)

# Step 6 : Display Results

result = pd.DataFrame({
    "Statistic": ["Mean", "Variance"],
    "Theoretical": [theoretical_mean, theoretical_variance],
    "Experimental": [
        round(experimental_mean,3),
        round(experimental_variance,3)
    ]
})

print("\nComparison Table\n")
print(result)

# Step 7 : Probability Distribution Plot

plt.figure(figsize=(10,6))

sns.histplot(
    heads,
    bins=np.arange(-0.5,n_flips+1.5,1),
    stat="probability",
    color="royalblue",
    edgecolor="black",
    kde=True
)

plt.axvline(
    experimental_mean,
    color="red",
    linestyle="--",
    linewidth=2,
    label=f"Mean = {experimental_mean:.2f}"
)

plt.title("Probability Distribution of Heads (15 Coin Flips)")
plt.xlabel("Number of Heads")
plt.ylabel("Probability")
plt.xticks(range(n_flips+1))
plt.legend()

plt.show()

# Step 8 : Probability Mass Function (PMF)

values, counts = np.unique(heads, return_counts=True)

pmf = counts / trials

plt.figure(figsize=(10,5))

plt.stem(values, pmf)

plt.title("Probability Mass Function (PMF)")
plt.xlabel("Number of Heads")
plt.ylabel("Probability")

plt.grid(alpha=0.3)

plt.show()

# Step 9 : Extra Analysis

print("\nHighest Heads Obtained :", np.max(heads))
print("Lowest Heads Obtained  :", np.min(heads))
print("Median Heads           :", np.median(heads))
print("Standard Deviation     :", round(np.std(heads),3))

# Step 10 : Frequency Table

frequency_table = pd.DataFrame({
    "Heads": values,
    "Frequency": counts,
    "Probability": np.round(pmf,4)
})

print("\nFrequency Table\n")
print(frequency_table)
```
